```python
#https://xgboost.readthedocs.io/en/release_3.2.0/tutorials/ 
#!pip install xgboost
import xgboost as xgb
```

```python
from sklearn.datasets import make_classification
num_classes = 3
X, y = make_classification(n_samples=1000, 
                           n_informative=5,
                           n_classes=num_classes)
```

```python
X.shape
```

#### Keep in mind

- When booster is set to gbtree or dart, XGBoost builds a tree model, which is a list of trees and can be sliced into multiple sub-models.



#### Training booster

```python
#internal data matrix optimized for efficiency and speed
dtrain = xgb.DMatrix(data=X, label=y)
```

```python
num_parallel_tree = 4
num_boost_round = 16
# total number of built trees is num_parallel_tree * num_classes * num_boost_round
```

```python
#We build a boosted random forest for classification here.
booster = xgb.train( {'num_parallel_tree': 4, 'subsample': 0.5, 'num_class': 3},
                    num_boost_round=16, 
                    dtrain=dtrain)
```

#### Exploring the config
